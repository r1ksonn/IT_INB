## Задание

В файле data_v2.txt записана длинная строка из латинских букв и цифр. Нужно посчитать, сколько символов в этой строке стоят между двумя четными цифрами.

Четные цифры – это 0, 2, 4, 6 и 8. Символ подходит, если слева от него стоит четная цифра и справа тоже стоит четная цифра. Например, в записи a2b4c буква b подходит, потому что слева от нее 2, а справа 4. А в записи 2x3 буква x не подходит, потому что 3 – нечетная цифра.

## Анализ задания, получение данных

Файл data_v2.txt лежит в репозитории преподавателя. Скачаем его в Google Colab такой командой:

```python
!wget https://raw.githubusercontent.com/enikolaev/IT-and-Programming/main/data/data_v2.txt
```

Файл можно загрузить и любым другим удобным способом.

Теперь прочитаем весь файл в одну строку. Чтение поместим в блок try/except: если файла вдруг нет, программа не упадет с ошибкой, а просто выведет понятное сообщение. Слово with нужно для того, чтобы файл сам закрылся после чтения.

```python
try:
  with open('data_v2.txt') as f:
    s = f.read().strip()
  print('Длина строки:', len(s))
except FileNotFoundError:
  print('Ошибка: файл data_v2.txt не найден')
  s = ''
```

Метод strip() убирает лишние пробелы и переносы строки в начале и в конце текста, чтобы они не мешали при подсчете.

Для проверки посмотрим на первые 20 символов строки и для каждого напишем, что это: буква, четная цифра или нечетная цифра.

```python
for i in range(20):
  c = s[i]
  if c.isdigit():
    if int(c) % 2 == 0:
      t = 'четная цифра'
    else:
      t = 'нечетная цифра'
  else:
    t = 'буква'
  print(c, t)
```

## Решение задачи

Идея простая: идем по строке и для каждого символа смотрим на его соседей слева и справа. Если оба соседа – четные цифры, прибавляем к счетчику единицу.

Первый и последний символы пропускаем, потому что у первого нет соседа слева, а у последнего нет соседа справа.

Чтобы не писать одну и ту же проверку два раза, сделаем небольшую функцию is_even_digit(). Она отвечает на вопрос, является ли символ четной цифрой.

```python
def is_even_digit(c):
  # Возвращает True, если символ c - четная цифра
  return c.isdigit() and int(c) % 2 == 0

count = 0
for i in range(1, len(s) - 1):
  if is_even_digit(s[i-1]) and is_even_digit(s[i+1]):
    count += 1

print('Количество символов между четными цифрами =', count)
```

Для наглядности посмотрим, сколько среди найденных символов букв и сколько цифр, и выведем несколько примеров.

```python
letters = 0
digits = 0
examples = []
for i in range(1, len(s) - 1):
  if is_even_digit(s[i-1]) and is_even_digit(s[i+1]):
    if s[i].isdigit():
      digits += 1
    else:
      letters += 1
    if len(examples) < 10:
      examples.append(s[i-1:i+2])

print('Из них букв:', letters)
print('Из них цифр:', digits)
print('Примеры троек:', examples)
```

Проверим ответ другим способом – с помощью регулярного выражения из модуля re. Оно ищет в строке все места, где идет четная цифра, потом любой символ, потом снова четная цифра. Если оба способа дадут одно и то же число, значит, ответ верный.

```python
import re
matches = re.findall(r'(?=[02468].[02468])', s)
print('Проверка через регулярное выражение:', len(matches))
```

В этой работе использовались условный оператор if, цикл for с range(), а также with и try/except для работы с файлом и обработки ошибок. Применялись функции open(), print(), len() и int(), методы строк read(), strip() и isdigit(), своя функция is_even_digit() и модуль re.
