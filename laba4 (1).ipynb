## Задание

В файле data_v2.txt посчитайте количество символов, расположенных между четными цифрами.

## Решение

```python
!wget https://raw.githubusercontent.com/enikolaev/IT-and-Programming/main/data/data_v2.txt
```

```python
try:
    text = open('data_v2.txt').read().strip()
except FileNotFoundError:
    print('Файл не найден')
    text = ''

even = '02468'
cnt = 0
for i in range(1, len(text) - 1):
    if text[i - 1] in even and text[i + 1] in even:
        cnt += 1
print(cnt)
```
