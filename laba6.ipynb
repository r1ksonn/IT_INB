## Задание

Текстовый файл pr6_1.txt состоит не более чем из 10^6 символов и содержит только заглавные буквы латинского алфавита и десятичные цифры. Найдите количество цифр, которые находятся между двумя буквами.

## Решение

```python
!wget https://raw.githubusercontent.com/enikolaev/IT-and-Programming/refs/heads/main/data/pract6/pr6_1.txt
```

```python
line = open('pr6_1.txt').read().strip()
k = 0
for i in range(1, len(line) - 1):
    if line[i].isdigit() and line[i - 1].isalpha() and line[i + 1].isalpha():
        k += 1
print(k)
```

## Ответ

4411
