## Задание

В файле pr5.txt содержится последовательность целых чисел. Элементы последовательности могут принимать целые значения от –10 000 до 10 000 включительно. Определите количество пар, в которых хотя бы один из двух элементов больше, чем среднее арифметическое всех чисел в файле, и хотя бы один из двух элементов делится на 17. В ответе запишите два числа: сначала количество найденных пар, а затем – максимальную сумму элементов таких пар. В данной задаче под парой подразумевается два идущих подряд элемента последовательности.

## Решение

```python
!wget https://raw.githubusercontent.com/enikolaev/IT-and-Programming/refs/heads/main/data/pract5/pr5.txt
```

```python
nums = [int(s) for s in open('pr5.txt')]
avg = sum(nums) / len(nums)
res = []
for i in range(len(nums) - 1):
    x, y = nums[i], nums[i + 1]
    if (x > avg or y > avg) and (abs(x) % 17 == 0 or abs(y) % 17 == 0):
        res.append(x + y)
print(len(res), max(res))
```

881 18941
