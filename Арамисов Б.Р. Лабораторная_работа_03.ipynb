## Задание

Напишите функцию, которая вычисляет и возвращает факториал
натурального числа, переданного функции в качестве параметра

## Решение

### Объявление функции

```python
def factorial(n):
    f = 1
    for i in range(1, n + 1):
        f *= i
    return f
```

### Демонстрация использования функции

```python
n = 5
print(f'Факториал числа {n} равен: {factorial(n)}')
n = 10
print(f'Факториал числа {n} равен: {factorial(n)}')
n = 1
print(f'Факториал числа {n} равен: {factorial(n)}')
```
